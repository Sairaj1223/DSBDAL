```python
import nltk
import pandas as pd
import requests
import string

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk import pos_tag
from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
from collections import Counter
from collections import Counter

# Download required NLTK data (IMPORTANT FIX)
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
nltk.download('wordnet')
```

```python
document = """
Data science is an interdisciplinary field.
It uses scientific methods and algorithms.
Machine learning is part of data science.
Data science helps in prediction and analysis.
"""
print(document)
```

```python
tokens = word_tokenize(document)

print("Tokens:")
print(tokens)
```

```python
stop_words = set(stopwords.words('english'))

filtered_tokens = [
    word for word in tokens
    if word.lower() not in stop_words and word.isalnum()
]

print("After Stopwords Removal:")
print(filtered_tokens)
```

```python
pos_tags = pos_tag(filtered_tokens)

print("POS Tags:")
print(pos_tags)
```

```python
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_tokens]

print("Stemmed Words:")
print(stemmed_words)
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_tokens]

print("Lemmatized Words:")
print(lemmatized_words)
```

```python
tf = Counter(filtered_tokens)

print("Term Frequency:")
print(tf)
```

```python
docs = [
    "Data science uses algorithms",
    "Machine learning is part of data science",
    "Data science helps prediction"
]

vectorizer = TfidfVectorizer(use_idf=True)

tfidf_matrix = vectorizer.fit_transform(docs)

idf_values = dict(zip(vectorizer.get_feature_names_out(), vectorizer.idf_))

print("IDF Values:")
print(idf_values)
```

```python
df_tfidf = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print("TF-IDF Matrix:")
print(df_tfidf)
```
