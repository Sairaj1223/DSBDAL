```python
summary = df.groupby('species')['sepal_length'].describe()
print("\nGrouped Summary:\n", summary)
```

```python
def custom_describe(x):
    return pd.Series({
        'count': x.count(),
        'mean': x.mean(),
        'std': x.std(),
        'mode': st.mode(x),
        'median': np.median(x),
        'min': x.min(),
        'max': x.max()
    })
```

```python
summary_with_mode_median = df.groupby('species')['sepal_length'].apply(custom_describe)
print("\nCustom Summary:\n", summary_with_mode_median)
```

```python
mode_summary = df.groupby('species')['sepal_length'].apply(lambda x: st.mode(x))
print("\nMode:\n", mode_summary)

median_summary = df.groupby('species')['sepal_length'].median()
print("\nMedian:\n", median_summary)

mean_summary = df.groupby('species')['sepal_length'].mean()
print("\nMean:\n", mean_summary)
```

```python
# ---------------- NUMERIC LIST ----------------
species_numeric_value = df.groupby('species').size().tolist()
print("\nSpecies count list:", species_numeric_value)
```

```python
# ---------------- FULL STATS ----------------
species_stats = df.groupby('species').describe()
print("\nAll statistics:\n", species_stats)
```

```python
# ---------------- FILTER BY SPECIES ----------------
print("\n--- Individual Species Statistics ---")

for sp in df['species'].unique():
    print(f"\n{sp} statistics:")
    print(df[df['species'] == sp].describe())
```
